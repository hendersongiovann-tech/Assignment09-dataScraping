## 4. Find all course records
Find all div elements with the course class. Print the number found. After your edit, you should find four.


```python
# TODO: Find the course containers and display their count.
```

```python
courses = soup.find_all("div", class_="course")
print(len(courses))
```

## 5. Extract names and credits
Loop through the course containers. For each container, extract its h2 text and the text of its credits paragraph. Convert the credit number to an integer. Build a list of dictionaries with the keys course_name and credits.

Keep each name paired with its own credit value. Do not type the course values directly into Python.


```python
# TODO: Extract all course records into a list of dictionaries.
```

```python
course_list = []

for course in courses:
    name = course.find("h2").get_text(strip=True)
    credit_text = course.find("p", class_="credits").get_text(strip=True)
    credits = int(credit_text.split()[0])

    course_list.append({
        "course_name": name,
        "credits": credits
    })

print(course_list)
```

## 6. Make a DataFrame and save a CSV
Create a DataFrame named courses_df with columns course_name and credits. Display it and save it as data/courses.csv without an index. Create the data folder if needed.


```python
# TODO: Build and display courses_df, then save the required CSV.
```

```python
courses_df = pd.DataFrame(course_list, columns=["course_name", "credits"])

display(courses_df)

Path("data").mkdir(exist_ok=True)

courses_df.to_csv("data/courses.csv", index=False)
```
